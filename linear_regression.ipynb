```python
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([40, 50, 60, 70, 80])
```

Screenshot 2026-06-09 105137.png

```python
model = LinearRegression()
model.fit(X, y)
```

```python
prediction = model.predict([[6]])
print(prediction)
```
