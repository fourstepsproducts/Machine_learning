```python
dataframe = read_csv("Machine_Learning/Classification/iris.csv")
dataframe.head()
```

```python
data = dataframe.drop('Variety', axis=1).values
y = dataframe['Variety'].values
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
y = encoder.fit_transform(y)
```

```python
model = DBSCAN(eps=0.6, min_samples=10)
model.fit(data)
```

```python
labels = model.labels_
print(labels)
```

```python
plt.scatter(data[:, 0], data[:, 1], c=labels, cmap='summer')
plt.title("DBSCAN Clusters")
plt.show()
```

```python
from sklearn.metrics import accuracy_score

print(accuracy_score(y, labels))
```
